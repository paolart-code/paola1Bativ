```python
import pygame
import random
import sys
import os

# Inicializa o Pygame
pygame.init()

LARGURA_TELA = 1000
ALTURA_TELA = 666
FPS = 60

COR_CACTO = (0, 128, 0)
COR_TEXTO = (255, 255, 255)

tela = pygame.display.set_mode((LARGURA_TELA, ALTURA_TELA))
pygame.display.set_caption("Jogo do Luis em São José dos Pinhais")
relogio = pygame.time.Clock()

# --- TAMANHO DO LUIS ---
LARGURA_LUIS = 130
ALTURA_LUIS = 145
ALTURA_LUIS_AGACHADO = 90

# --- LÓGICA PORTÁTIL (PEGA A PASTA ONDE O CÓDIGO ESTÁ RODANDO) ---
PASTA_DO_JOGO = os.path.dirname(os.path.abspath(__file__))

def achar_arquivo_local(nome_base):
    for arquivo in os.listdir(PASTA_DO_JOGO):
        if arquivo.lower().startswith(nome_base.lower()):
            return os.path.join(PASTA_DO_JOGO, arquivo)
    return None

# Carrega Fundo
caminho_fundo = achar_arquivo_local("fundo")
if caminho_fundo:
    imagem_fundo = pygame.image.load(caminho_fundo)
    imagem_fundo = pygame.transform.scale(imagem_fundo, (LARGURA_TELA, ALTURA_TELA))
else:
    print(/content/1000063862.png)
    pygame.quit()
    sys.exit()

def carregar_e_remover_fundo(caminho, larg, alt):
    img = pygame.image.load(caminho).convert()
    img.set_colorkey((0, 0, 0))
    return pygame.transform.scale(img, (larg, alt))

# --- CARREGA AS ANIMAÇÕES NOVAS SEGUINDO A LÓGICA (luis1, luis2, luis3) ---
sprites_corrida = []
for i in range(1, 4):
    caminho = achar_arquivo_local(f"luis{i}")
    if caminho:
        sprites_corrida.append(carregar_e_remover_fundo(caminho, LARGURA_LUIS, ALTURA_LUIS))

if not sprites_corrida:
    caminho_padrao = achar_arquivo_local("luis")
    if caminho_padrao:
        sprites_corrida = [carregar_e_remover_fundo(caminho_padrao, LARGURA_LUIS, ALTURA_LUIS)]
    else:
        print(/content/1000063862.png)
        pygame.quit()
        sys.exit()

caminho_pulo = achar_arquivo_local("luis_pulo")
sprite_pulo = carregar_e_remover_fundo(caminho_pulo, LARGURA_LUIS, ALTURA_LUIS) if caminho_pulo else sprites_corrida[0]

caminho_agacha = achar_arquivo_local("luis_agacha")
sprite_agacha = carregar_e_remover_fundo(caminho_agacha, LARGURA_LUIS, ALTURA_LUIS_AGACHADO) if caminho_agacha else sprites_corrida[0]


class Luis:
    def __init__(self):
        self.largura = LARGURA_LUIS
        self.altura_normal = ALTURA_LUIS
        self.altura_agachado = ALTURA_LUIS_AGACHADO
        self.x = 80
        self.y_chao = ALTURA_TELA - self.altura_normal - 60
        self.y = self.y_chao

        self.velocidade_y = 0
        self.gravidade = 1.3
        self.esta_pulando = False
        self.esta_agachado = False

        self.index_animacao = 0
        self.rect = pygame.Rect(self.x, self.y, self.largura, self.altura_normal)

    def pular(self):
        if not self.esta_pulando and not self.esta_agachado:
            self.velocidade_y = -22
            self.esta_pulando = True

    def agachar(self, apertado):
        if not self.esta_pulando:
            self.esta_agachado = apertado
            if self.esta_agachado:
                self.rect.height = self.altura_agachado
                self.y = self.y_chao + (self.altura_normal - self.altura_agachado)
            else:
                self.rect.height = self.altura_normal
                self.y = self.y_chao

    def atualizar(self):
        if self.esta_pulando:
            self.velocidade_y += self.gravidade
            self.y += self.velocidade_y
            if self.y >= self.y_chao:
                self.y = self.y_chao
                self.velocidade_y = 0
                self.esta_pulando = False

        self.rect.y = self.y

        # Avança a animação frame por frame
        self.index_animacao += 0.15
        if self.index_animacao >= len(sprites_corrida):
            self.index_animacao = 0

    def desenhar(self, superficie):
        if self.esta_pulando:
            superficie.blit(sprite_pulo, (self.x, self.y))
        elif self.esta_agachado:
            superficie.blit(sprite_agacha, (self.x, self.y))
        else:
            frame_atual = sprites_corrida[int(self.index_animacao)]
            superficie.blit(frame_atual, (self.x, self.y))


class Cactus:
    def __init__(self):
        self.largura = 30
        self.altura = random.randint(50, 95)
        self.x = LARGURA_TELA + random.randint(0, 300)
        self.y = ALTURA_TELA - self.altura - 60
        self.velocidade = 8.5
        self.rect = pygame.Rect(self.x, self.y, self.largura, self.altura)

    def atualizar(self):
        self.x -= self.velocidade
        self.rect.x = self.x

    def desenhar(self, superficie):
        pygame.draw.rect(superficie, COR_CACTO, self.rect)
```
